```python
import json

try:
    with open("inventory.json", "r") as file:
        inventory = json.load(file)

except FileNotFoundError:
    inventory = []
print(inventory)
```

```python
def save_records():
    with open("inventory.json", "w") as file:
        json.dump(inventory, file, indent=4)
```

```python
def add_product():
    product_id = input("Enter product ID: ")

    for product in inventory:
        if product["id"] == product_id:
            print("Product ID already exists!")
            return

    name = input("Enter product name: ")
    price = float(input("Enter product price: "))
    quantity = int(input("Enter product quantity: "))

    product = {
        "id": product_id,
        "name": name,
        "price": price,
        "quantity": quantity
    }

    inventory.append(product)
    save_records()

    print("Product added successfully!")

```

```python
def update_product():
    product_id = input("Enter product ID to update: ")

    for product in inventory:
        if product["id"] == product_id:

            print("Product found!")
            print("Current Name:", product["name"])
            print("Current Price:", product["price"])
            print("Current Quantity:", product["quantity"])

            new_price = float(input("Enter new price: "))
            new_quantity = int(input("Enter new quantity: "))

            product["price"] = new_price
            product["quantity"] = new_quantity

            save_records()

            print("Product updated successfully!")
            return

    print("Product not found!")
```

```python
def delete_product():
    product_id = input("Enter product ID to delete: ")

    for product in inventory:
        if product["id"] == product_id:
            inventory.remove(product)
            save_records()

            print("Product deleted successfully!")
            return

    print("Product not found!")
```

```python
def search_product():
    search = input("Enter product ID or product name: ").lower()

    for product in inventory:
        if product["id"].lower() == search or product["name"].lower() == search:
            print("\nProduct Found!")
            print("Product ID:", product["id"])
            print("Product Name:", product["name"])
            print("Price:", product["price"])
            print("Quantity:", product["quantity"])
            return

    print("Product not found!")
```

```python
def inventory_report():
    total_products = len(inventory)
    total_quantity = 0
    total_value = 0

    for product in inventory:
        total_quantity += product["quantity"]
        total_value += product["price"] * product["quantity"]

    print("\n===== INVENTORY REPORT =====")
    print("Total Products:", total_products)
    print("Total Quantity:", total_quantity)
    print("Total Inventory Value:", total_value)

    print("\n===== LOW STOCK PRODUCTS =====")

    low_stock_found = False

    for product in inventory:
        if product["quantity"] < 10:
            print(
                product["id"],
                "-",
                product["name"],
                "| Stock:",
                product["quantity"]
            )
            low_stock_found = True

    if not low_stock_found:
        print("No low-stock products.")
```

```python
inventory_report()
```

```python
def main_menu():
    while True:
        print("\n===== INVENTORY MANAGEMENT SYSTEM =====")
        print("1. Add Product")
        print("2. Update Product")
        print("3. Delete Product")
        print("4. Search Product")
        print("5. Inventory Report")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_product()

        elif choice == "2":
            update_product()

        elif choice == "3":
            delete_product()

        elif choice == "4":
            search_product()

        elif choice == "5":
            inventory_report()

        elif choice == "6":
            print("Thank you for using Inventory Management System!")
            break

        else:
            print("Invalid choice! Please try again.")
```
